```python
# From Brooke Digia's fried function in maos_utils as of 6/9/26
# r0z1 = 0.976 * ( wvl.to(u.m) / arcsec_to_rad(DIMM).to('', equivalencies=u.dimensionless_angles() ))

import numpy as np
import matplotlib.pyplot as plt
import astropy.units as u


wvl = 500 * u.nm # Default
DIMM_array = np.array([0.37, 0.21, 0.8]) * u.arcsec

def arcsec_to_rad(arcsec):
    return arcsec.to(u.rad, equivalencies=u.dimensionless_angles())

def r0(DIMM):
    return 0.976 * (wvl.to(u.m) / arcsec_to_rad(DIMM).to('', equivalencies=u.dimensionless_angles()))

r0_values = r0(DIMM_array)
print(r0_values)
```
